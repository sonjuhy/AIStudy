```python
import torch

b, c, h, w = 1, 1, 2, 2
d4_channel = torch.rand(b, c, h, w)
print(d4_channel)
print(d4_channel.shape)
```

```python
viewed_channel = d4_channel.view(b, h * w, c)
print(viewed_channel)
print(viewed_channel.shape)
```

```python
permuted_channel = viewed_channel.permute(0, 2, 1)
print(permuted_channel)
print(permuted_channel.shape)
```

```python
permuted_channel = viewed_channel.permute(0, 2, 1)
print(permuted_channel)
print(permuted_channel.shape)
```

```python
flatten_channel = torch.flatten(permuted_channel, start_dim=0)
print(flatten_channel)
print(flatten_channel.shape)
```
